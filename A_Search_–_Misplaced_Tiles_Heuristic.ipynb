```python

import heapq

initial = (5, 4, 6,
           1, 8, 7,
           3, 2, 0)

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)


# Misplaced Tiles Heuristic
def heuristic(state):
    count = 0

    for i in range(9):
        if state[i] != 0 and state[i] != goal[i]:
            count += 1

    return count


# Generate Neighbors
def get_neighbors(state):
    neighbors = []
    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_pos = r * 3 + c
            new_state = list(state)

            new_state[pos], new_state[new_pos] = (
                new_state[new_pos], new_state[pos])

            neighbors.append(tuple(new_state))

    return neighbors


# A* Search
def astar(initial, goal):
    pq = [(heuristic(initial), 0, initial)]
    visited = set()
    parent = {initial: None}
    cost = {initial: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)

        print("g(n):", g, "h(n):", heuristic(current),
              "f(n):", f)

        if current == goal:
            print("\nGoal Reached!")

            path = []
            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            for state in path:
                for i in range(0, 9, 3):
                    print(state[i:i+3])
                print()

            print("Total Moves:", len(path) - 1)
            return

        for neighbor in get_neighbors(current):
            new_g = g + 1

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = current
                new_f = new_g + heuristic(neighbor)

                heapq.heappush(
                    pq, (new_f, new_g, neighbor))

    print("No Solution Found")


astar(initial, goal)
```
