```python

# Simple Reflex Agent

def simple_reflex_agent():
    rooms = {
        "A": "Dirty",
        "B": "Dirty"
    }

    location = "A"

    print("\nSIMPLE REFLEX AGENT")

    while "Dirty" in rooms.values():
        print("\nCurrent Location:", location)
        print("Room Status:", rooms)

        if rooms[location] == "Dirty":
            print("Room", location, "is Dirty. Cleaning...")
            rooms[location] = "Clean"

        else:
            if location == "A":
                location = "B"
            else:
                location = "A"

            print("Moving to Room", location)

    print("\nAll rooms are clean!")
    print("Final Status:", rooms)


# Goal-Based Agent

def goal_based_agent():
    rooms = {
        "A": "Dirty",
        "B": "Dirty"
    }

    location = "A"
    goal = "All rooms Clean"

    print("\nGOAL-BASED AGENT")
    print("Goal:", goal)

    while rooms["A"] != "Clean" or rooms["B"] != "Clean":

        print("\nCurrent Location:", location)

        if rooms[location] == "Dirty":
            print("Cleaning Room", location)
            rooms[location] = "Clean"

        elif location == "A":
            location = "B"
            print("Moving to Room B")

        elif location == "B":
            location = "A"
            print("Moving to Room A")

    print("\nGoal Achieved!")
    print("Final Status:", rooms)


simple_reflex_agent()
goal_based_agent()
```
