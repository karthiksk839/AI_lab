```python

import heapq

initial = (2, 8, 3,
           1, 6, 4,
           7, 5, 0)

goal = (1, 2, 3,
        8, 0, 4,
        7, 6, 5)


# Manhattan Distance Heuristic
def heuristic(state):
    distance = 0

    for i in range(9):
        if state[i] != 0:
            row, col = divmod(i, 3)
            goal_pos = goal.index(state[i])
            goal_row, goal_col = divmod(goal_pos, 3)

            distance += abs(row - goal_row)
            distance += abs(col - goal_col)

    return distance


# Generate Neighbors
def get_neighbors(state):
    neighbors = []
    pos = state.index(0)
    row, col = divmod(pos, 3)

    moves = [(-1, 0), (1, 0), (0, -1), (0, 1)]

    for dr, dc in moves:
        r, c = row + dr, col + dc

        if 0 <= r < 3 and 0 <= c < 3:
            new_pos = r * 3 + c
            new_state = list(state)

            new_state[pos], new_state[new_pos] = (
                new_state[new_pos], new_state[pos])

            neighbors.append(tuple(new_state))

    return neighbors


# A* Search
def astar(initial, goal):
    pq = [(heuristic(initial), 0, initial)]
    visited = set()
    parent = {initial: None}
    cost = {initial: 0}

    while pq:
        f, g, current = heapq.heappop(pq)

        if current in visited:
            continue

        visited.add(current)

        print("g(n):", g, "h(n):", heuristic(current),
              "f(n):", f)

        if current == goal:
            print("\nGoal Reached!")

            path = []
            while current is not None:
                path.append(current)
                current = parent[current]

            path.reverse()

            for state in path:
                for i in range(0, 9, 3):
                    print(state[i:i+3])
                print()

            print("Total Moves:", len(path) - 1)
            return

        for neighbor in get_neighbors(current):
            new_g = g + 1

            if neighbor not in cost or new_g < cost[neighbor]:
                cost[neighbor] = new_g
                parent[neighbor] = current
                new_f = new_g + heuristic(neighbor)

                heapq.heappush(
                    pq, (new_f, new_g, neighbor))

    print("No Solution Found")


astar(initial, goal)
```
