```python

import random
import math

# Calculate attacking queen pairs
def calculate_cost(board):
    cost = 0

    for i in range(8):
        for j in range(i + 1, 8):
            if board[i] == board[j] or abs(board[i] - board[j]) == abs(i - j):
                cost += 1

    return cost


# Simulated Annealing algorithm
def simulated_annealing():

    # Random initial arrangement
    board = [random.randint(0, 7) for i in range(8)]

    temperature = 100
    cooling_rate = 0.99

    while temperature > 0.1:

        current_cost = calculate_cost(board)

        if current_cost == 0:
            return board

        # Generate neighbouring state
        new_board = board.copy()
        col = random.randint(0, 7)
        new_board[col] = random.randint(0, 7)

        new_cost = calculate_cost(new_board)

        # Accept better or probabilistically accept worse state
        if new_cost < current_cost:
            board = new_board
        else:
            probability = math.exp((current_cost - new_cost) / temperature)

            if random.random() < probability:
                board = new_board

        # Reduce temperature
        temperature *= cooling_rate

    return board


# Run the program
solution = simulated_annealing()

print("8 Queens Solution:")

for row in range(8):
    for col in range(8):
        if solution[col] == row:
            print("Q", end=" ")
        else:
            print(".", end=" ")
    print()

print("Number of attacking pairs:", calculate_cost(solution))
```
