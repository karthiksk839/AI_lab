```python
# Forward Chaining Algorithm

facts = {
    "American(Robert)",
    "Enemy(A,America)",
    "Owns(A,T1)",
    "Missile(T1)"
}

rules = [
    (["Missile(T1)"], "Weapon(T1)"),
    (["Enemy(A,America)"], "Hostile(A)"),
    (["Missile(T1)", "Owns(A,T1)"], "Sells(Robert,T1,A)"),
    (["American(Robert)", "Weapon(T1)",
      "Sells(Robert,T1,A)", "Hostile(A)"], "Criminal(Robert)")
]

def forward_chaining(facts, rules, goal):

    while True:
        new_facts = []

        for conditions, result in rules:
            if all(condition in facts for condition in conditions):
                if result not in facts:
                    new_facts.append(result)

                    if result == goal:
                        print("New Fact:", result)
                        return True

        if not new_facts:
            return False

        for fact in new_facts:
            facts.add(fact)
            print("New Fact:", fact)


goal = "Criminal(Robert)"

print("Initial Facts:")
for fact in facts:
    print(fact)

if forward_chaining(facts, rules, goal):
    print("\nGoal Proved:", goal)
else:
    print("\nGoal Cannot Be Proved")
```
