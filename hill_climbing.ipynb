```python

import random

N = 4
random.seed(42)

# Calculate the number of attacking queen pairs
def calculate_cost(state):
    cost = 0

    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j] or abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


# Generate all neighboring states
def get_neighbors(state):
    neighbors = []

    for col in range(N):
        for row in range(N):
            if row != state[col]:
                new_state = state.copy()
                new_state[col] = row
                neighbors.append(new_state)

    return neighbors


# Display chessboard
def display(state):
    for row in range(N):
        for col in range(N):
            if state[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


# Hill Climbing Algorithm
def hill_climbing():

    current = [0, 0, 0, 0]
    step = 1

    while True:

        current_cost = calculate_cost(current)

        print("\nStep:", step)
        print("Current State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("\nGoal Configuration Reached!")
            display(current)
            return

        neighbors = get_neighbors(current)

        print("\nNeighbor States and Costs:")

        best_cost = float('inf')
        best_neighbors = []

        for neighbor in neighbors:
            cost = calculate_cost(neighbor)

            print(neighbor, "-> Cost:", cost)

            if cost < best_cost:
                best_cost = cost
                best_neighbors = [neighbor]

            elif cost == best_cost:
                best_neighbors.append(neighbor)

        # Choose the best neighbor
        if best_cost < current_cost:
            current = random.choice(best_neighbors)
            print("\nMoving to:", current)

        else:
            print("\nLocal Minimum Reached. Restarting...")
            current = [random.randint(0, N - 1) for _ in range(N)]

        step += 1


# Main Program
hill_climbing()
```
