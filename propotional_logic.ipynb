```python

from itertools import product

# Truth table enumeration algorithm

def implies(p, q):
    return not p or q


def model_check(Q, P, R):
    # Knowledge Base
    kb1 = implies(Q, P)
    kb2 = implies(P, not Q)
    kb3 = Q or R

    KB = kb1 and kb2 and kb3

    # Queries
    query1 = R
    query2 = implies(R, P)
    query3 = implies(Q, R)

    return KB, query1, query2, query3


# Generate all possible truth combinations
print("Q\tP\tR\tQ->P\tP->~Q\tQ OR R\tKB")

models = []

for Q, P, R in product([True, False], repeat=3):

    kb1 = implies(Q, P)
    kb2 = implies(P, not Q)
    kb3 = Q or R

    KB = kb1 and kb2 and kb3

    print(Q, P, R, kb1, kb2, kb3, KB, sep="\t")

    if KB:
        models.append((Q, P, R))


# Check entailment
print("\nModels where KB is TRUE:")
for model in models:
    print(model)

print("\nEntailment Results:")

queries = {
    "R": lambda q, p, r: r,
    "R -> P": lambda q, p, r: implies(r, p),
    "Q -> R": lambda q, p, r: implies(q, r)
}

for name, query in queries.items():

    result = all(query(q, p, r) for q, p, r in models)

    print(name, ":", "ENTAILED" if result else "NOT ENTAILED")
```
