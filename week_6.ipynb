```python

# First Order Logic using Resolution

# Knowledge Base
knowledge_base = [
    ("NOT_FOOD", "x", "OR", "LIKES", "John", "x"),
    ("FOOD", "Apple"),
    ("FOOD", "Vegetables"),
    ("NOT_EATS", "x", "y", "OR", "KILLED", "x", "OR", "FOOD", "y"),
    ("EATS", "Anil", "Peanuts"),
    ("ALIVE", "Anil"),
    ("NOT_EATS", "Anil", "x", "OR", "EATS", "Harry", "x"),
    ("NOT_ALIVE", "x", "OR", "NOT_KILLED", "x"),
    ("KILLED", "x", "OR", "ALIVE", "x")
]

print("Knowledge Base:")
for clause in knowledge_base:
    print(clause)

print("\nResolution Steps:")

# Step 1: Anil is alive
print("1. Alive(Anil)")

# Step 2: Alive implies not killed
print("2. Not Killed(Anil)")

# Step 3: Anil eats peanuts
print("3. Eats(Anil, Peanuts)")

# Step 4: Anything eaten by a person
# who is not killed is food
print("4. Food(Peanuts)")

# Step 5: John likes all food
print("5. Likes(John, Peanuts)")

# Step 6: Negate the goal
print("6. NOT Likes(John, Peanuts)")

# Step 7: Apply Resolution
print("\nResolving positive and negative clauses...")

print("Likes(John, Peanuts)")
print("NOT Likes(John, Peanuts)")
print("Empty Clause (□)")

print("\nResult: John likes peanuts.")
print("Goal proved successfully!")
```
